```python
import re

def clean_newlines(text):
    if not isinstance(text, str):
        return text



    text=text.replace("\xa0"," ")

    # Fix mid-sentence splits.

    text = re.sub(r"(?<=[a-zA-Z0-9,;])\n(?=[a-zA-Z0-9])", " ", text)

    # Collapse excessive blank lines.

    text = re.sub(r"(\s*\n\s*){3,}", "\n", text)

    # Standardize any remaining stray double spaces back down to single spaces
    text = re.sub(r"[ \t]+", " ", text)

     # Strip leading and trailing whitespaces/newlines
    text = text.strip()

    return text


df = df.dropna(subset=["Context", "Response"])

df["Context"] = df["Context"].apply(clean_newlines)
df["Response"] = df["Response"].apply(clean_newlines)

print("Cleaned Context")
print(df["Context"].iloc[0])
print("\nResponse Example")
print(df["Response"].iloc[0])
```

```python
def remove_forum_noise(text):
    if not isinstance(text, str):
        return text

    #  Remove URLs
    url_pattern = r"https?://\S+|www\.\S+|(?:\w+\.)+(?:com|org|net|edu)\/\S*"
    text = re.sub(url_pattern, "", text)

    # Remove common text-based email address patterns
    email_pattern = r"\S+@\S+\.\S+"
    text = re.sub(email_pattern, "", text)

    # Strip professional titles commonly found in signatures

    credentials = r",?\s*(?:LCSW|LMFT|LPC|PsyD|PhD|Ph\.D\.|M\.D\.|MD|MSW|NCC)\b"
    text = re.sub(credentials, "", text, flags=re.IGNORECASE)

    # Clean up generic boilerplate signature sign-offs at the end of responses

    sign_offs = [
        r"(?i)best regards,?\s*$",
        r"(?i)warmly,?\s*$",
        r"(?i)sincerely,?\s*$",
        r"(?i)hope this helps,?\s*$",
        r"(?i)good luck to you\.?\s*$",
    ]
    for pattern in sign_offs:
        text = re.sub(pattern, "", text)

    text = text.strip()

    text = re.sub(r"[,;\-\s]+$", ".", text)

    return text


df["Response"] = df["Response"].apply(remove_forum_noise)

df["Context"] = df["Context"].apply(remove_forum_noise)


print("Cleaned Response")
print(df["Response"].iloc[0])
```

```python
def convert_to_chat_format(row):
    #  system prompt
    system_message = (
        "You are an AI counseling assistant. Your role is to provide a warm, empathetic, "
        "and non-judgmental space. Listen actively, validate emotions, and guide users "
        "using open-ended questions. Never provide psychiatric diagnoses or medication dosages."
    )

    # standard messages format
    return [
        {"role": "system", "content": system_message},
        {"role": "user", "content": row["Context"]},
        {"role": "assistant", "content": row["Response"]},
    ]


df["messages"] = df.apply(convert_to_chat_format, axis=1)


print("Structured Training Message")
import pprint

pprint.pprint(df["messages"].iloc[0])
```

```python
from datasets import Dataset

final_df = df[["messages"]]

hf_train_dataset = Dataset.from_pandas(final_df)
print(hf_train_dataset)
```

```python
import torch
from datasets import load_dataset
from transformers import TrainingArguments
from unsloth import FastLanguageModel
from unsloth.chat_templates import get_chat_template, train_on_responses_only
from trl import SFTTrainer

# Tighten max sequence length to reduce attention matrix load
max_seq_length = 1024
dtype = None
load_in_4bit = True

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/Qwen2.5-7B-Instruct-bnb-4bit",
    max_seq_length=max_seq_length,
    dtype=dtype,
    load_in_4bit=load_in_4bit,
)

model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    target_modules=[
        "q_proj",
        "k_proj",
        "v_proj",
        "o_proj",
        "gate_proj",
        "up_proj",
        "down_proj",
    ],
    lora_alpha=16,
    lora_dropout=0,
    bias="none",
    use_gradient_checkpointing="unsloth",  # Keeps memory usage locked down
    random_state=3407,
)

tokenizer = get_chat_template(
    tokenizer,
    chat_template="qwen-2.5",
)


def formatting_prompts_func(examples):
    convos = examples["messages"]
    texts = [
        tokenizer.apply_chat_template(
            convo, tokenize=False, add_generation_prompt=False
        )
        for convo in convos
    ]
    return {"text": texts}


unsloth_dataset = hf_train_dataset.map(formatting_prompts_func, batched=True)

trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=unsloth_dataset,
    dataset_text_field="text",
    max_seq_length=max_seq_length,
    dataset_num_proc=2,
    packing=False,
    args=TrainingArguments(
        #  Lower absolute batch size and raise accumulation steps
        per_device_train_batch_size=1,
        gradient_accumulation_steps=8,
        warmup_steps=5,
        max_steps=60,
        learning_rate=2e-4,
        fp16=not torch.cuda.is_bf16_supported(),
        bf16=torch.cuda.is_bf16_supported(),
        logging_steps=1,
        output_dir="outputs",
        # Enforce 8-bit optimizer allocation to save ~30% VRAM
        optim="adamw_8bit",
    ),
)

trainer = train_on_responses_only(
    trainer,
    instruction_part="<|im_start|>user\n",
    response_part="<|im_start|>assistant\n",
)

# Clear hidden PyTorch memory caches
torch.cuda.empty_cache()

trainer.train()
```

```python
# Switch the model into fast inference mode
FastLanguageModel.for_inference(model)

# new test prompt
test_conversation = [
    {
        "role": "system",
        "content": "You are an AI counseling assistant. Your role is to provide a warm, empathetic, and non-judgmental space.",
    },
    {
        "role": "user",
        "content": "I've been feeling incredibly overwhelmed with my university classes lately, and it feels like I'm just going to fail everything.",
    },
]

# Format the input scenario using the model's tokenizer
inputs = tokenizer.apply_chat_template(
    test_conversation,
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt",
).to("cuda")

#  Generate the response text
outputs = model.generate(
    input_ids=inputs,
    max_new_tokens=150,  # Limits response length
    use_cache=True,
    temperature=0.7,  # Controls creativity
    top_p=0.9,
)

# Decode and display the therapist's response
decoded_response = tokenizer.batch_decode(outputs)
print("--- AI Therapist Response Test ---")
print(decoded_response[0])
```

```python
#  Clean up generation settings
FastLanguageModel.for_inference(model)

test_prompt = [
    {
        "role": "system",
        "content": "You are an AI counseling assistant. Your role is to provide a warm, empathetic, and non-judgmental space.",
    },
    {
        "role": "user",
        "content": "I've been feeling incredibly overwhelmed with my university classes lately, and it feels like I'm just going to fail everything.",
    },
]

# Tokenize the input text cleanly with attention masks included
inputs = tokenizer.apply_chat_template(
    test_prompt,
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt",
    return_dict=True,  # Packs input_ids and attention_mask together cleanly
).to("cuda")


# Calling .base_model
untuned_outputs = model.base_model.generate(
    **inputs, max_new_tokens=150, temperature=0.7, use_cache=True
)


print("   ORIGINAL UNTUNED BASE MODEL RESPONSE")

print(tokenizer.batch_decode(untuned_outputs)[0])



# Calling the model normally
tuned_outputs = model.generate(
    **inputs, max_new_tokens=150, temperature=0.7, use_cache=True
)

print("  FINE-TUNED THERAPIST MODEL RESPONSE")

print(tokenizer.batch_decode(tuned_outputs)[0])
```
